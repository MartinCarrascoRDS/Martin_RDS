# Feriados según la librería `holidays`

Este notebook genera un DataFrame con los feriados oficiales de Chile y Perú, junto con la descripción asociada a cada uno, usando la librería `holidays`.

```python
import pandas as pd
import holidays

# Definir año a analizar (puedes cambiarlo libremente)
anio = 2026

# Crear objeto de feriados para Chile
feriados_cl = holidays.CL(years=[anio])

# Convertir a DataFrame
df_feriados = pd.DataFrame([
    {'FECHA': fecha, 'FERIADO': nombre}
    for fecha, nombre in feriados_cl.items()
])

# Ordenar por fecha
df_feriados['FECHA'] = pd.to_datetime(df_feriados['FECHA'])
df_feriados = df_feriados.sort_values('FECHA').reset_index(drop=True)

df_feriados
```

## Notas
- La columna **FERIADO** corresponde exactamente al nombre entregado por la librería `holidays`.
- Si un feriado cambia de fecha por ley o decreto, la librería lo ajusta automáticamente.
- Puedes usar este DataFrame directamente para cruces con ventas o clasificación de días.

```python
feriados_pe = holidays.PE(years=[anio])

df_feriados_pe = pd.DataFrame([
    {'FECHA': fecha, 'FERIADO': nombre}
    for fecha, nombre in feriados_pe.items()
])

df_feriados_pe['FECHA'] = pd.to_datetime(df_feriados_pe['FECHA'])
df_feriados_pe = df_feriados_pe.sort_values('FECHA').reset_index(drop = True)

df_feriados_pe
```
